A place where I look into the data to see if there's any patterns/for inspiration on how to forecasts.

```python
# imports
import pandas as pd
import pyarrow.parquet as pq
```

```python
# loading data
# Read the Parquet file
table = pq.read_table("datasets/train.parquet")

# Convert to a pandas DataFrame
df = table.to_pandas()

print("First five data entries:")
print(df.head())

df
```

```python
N = df.shape[1] - 3
FEATURE_COLUMNS = [f"{i}" for i in range(N)]
#print(FEATURE_COLUMNS)

df[FEATURE_COLUMNS]
```

```python
# selecting based on seq_ix
curr_seq_ix = 0

filtered_df = df.loc[df['seq_ix'] == curr_seq_ix]

# output first 10 and last 10 rows to check if they need prediction
print(filtered_df.iloc[99:110, 2])
print(filtered_df.iloc[-10:, 2])
```
